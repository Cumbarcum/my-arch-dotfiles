## Необходимо реализовать машину конечных состояний (FSM) для управления абстрактным спутником.

## Спутник должен иметь 5 состояний:
- IDLE - ожидание команд
- SCAN - сбор данных
- TRANSMIT - передача данных
- CHARGE - зарядка
- MANEUVER - перемещение

## Спутник распоряжается следующими ресурсами:
- data - данные. Спутник может нести максимум 10 data.
- charge - заряд, необходим для сканирования и передачи данных. Восстанавливается зарядкой. Максимум 10.
- fuel - топливо, необходимое для манёвров. Не восстанавливается. Начальное количество 10.

## Условия:
- Из IDLE можно попасть в любое состояние.
- Каждый переход между состояниями тратит 1 заряд.
- Ошибка при переходе приводит в состояние IDLE и тоже тратит 1 заряд.
- SCAN собирает 5 данных если предыдущее действие не было MANEUVER, иначе 2.
- SCAN может собрать не более 5 данных с одного места, т.е. он должен переместиться, чтобы собрать новые данные.
- Если данных SCAN больше объема памяти, спутник может собрать их излишек после передачи данных.
- SCAN потребляет 2 заряда.
- TRANSMIT передаёт все данные спутника и очищает память.
- Нельзя перейти в TRANSMIT если данных нет.
- Нельзя перейти в TRANSMIT после MANEUVER.
- TRANSMIT тратит 4 заряда.
- MANEUVER потребляет 1 топливо.
- В MANUEVER нельзя перейти сразу после CHARGE.
- CHARGE восстанавливает 3 заряда.
- Переход В/ИЗ состояние CHARGE НЕ тратит заряд.
- Нельзя выполнять CHARGE больше 2 раз подряд. IDLE между двумя CHARGE не сбрасывает это условие.
- Если после CHARGE заряд превышает максимальный предел, спутник тратит 1 топливо на охлаждение.
