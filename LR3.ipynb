### Теория: Метод трапеций
Для поиска численного значения интеграла необходимо разбить площадь под графиком функции на $n$ вертиакльных участков шириной $h$.

В методе трапеций верхняя граница каждого участка аппроксимируется прямой линией (хордой), соединяющей точки графика. Таким образом, каждый участок превращается в прямоугольную трапецию.

- Ширина (шаг) каждой трапеции: $$h=\frac{b-a}{n},$$
где $a$ и $b$ - нижний и верхний предел интегрирования соответственно.
- Площадь $i$-й трапеции:
$$S_{i}=\frac{f(x_{i})+f(x_{i+1})}{2}\cdot h$$


Если сложить площади всех трапеций, то значения функции во внутренних точках сложатся дважды, а крайние точки ($a$ и $b$) — по одному разу. Получается итоговая составная формула трапеций:
$$I\approx h\cdot \left(\frac{f(a)+f(b)}{2}+\sum _{i=1}^{n-1}f(x_{i})\right),$$
где $x_i = a + i \cdot h$ — координаты внутренних узлов разбиения.

## Теория: Метод дихотомии

Если непрерывная функция $f(x)$ на концах отрезка $[a, b]$ имеет разные знаки (то есть $f(a) \cdot f(b) < 0$), то внутри этого отрезка существует как минимум один корень уравнения $f(x) = 0$.
Алгоритм метода:
1. Найти середину текущего отрезка: $c = \frac{a + b}{2}$.
2. Вычислить значение функции $f(c)$.
3. Если $f(c) \approx 0$ (с заданной точностью $\varepsilon$) или длина отрезка $(b - a)$ стала меньше $\varepsilon$, то корень найден.
4. Если корень не найден, выбирается та половина отрезка, на концах которой функция меняет знак:
  - Если $f(a) \cdot f(c) < 0$, то корень лежит в левой половине. *Правой* границей становится $c$, т.е. $b = c$.
  - В противном случае корень лежит в правой половине. *Левой* границей становится $c$, т.е. $a = c$.
5. Повторять до тех пор, пока не найден корень.
